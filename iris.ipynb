```python
x = df.drop(columns =(['Id',"Species"]))
```

```python
x
```

```python
y = df['Species']
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,train_size=0.8,random_state=56)
```

```python
x_train
```

```python
x_test
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
knn = KNeighborsClassifier()
```

```python
model = knn.fit(x_train,y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
```

```python
accuracy_score(y_test,y_pred)
```

```python
cr = classification_report(y_test,y_pred)
print(cr)
```

```python
cm = confusion_matrix(y_test,y_pred)
cm
```

```python
import joblib
```

```python
model = joblib.dump(model,"iris_model.pkl")
```

```python
#deployment
import streamlit as st
import numpy as np
```

```python
#load model
model = joblib.load("iris_model.pkl")
```

```python
#page title
st.title("machine learning on iris dataset")
```

```python
#input labels
sepal_length = st.number_input("sepal_length")
sepal_width = st.number_input("sepal_width")
petal_length = st.number_input("petal_length")
petal_width = st.number_input("petal_width")
```

```python
#prediction
if st.button("predict"):
  input = np.array([[sepal_length,
                     sepal_width,
                     petal_length,
                     petal_width]]).astype(np.float64)
  prediction = model.predict(input)
  st.success(prediction)
```
